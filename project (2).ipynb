```python
# Task 6: Univariate Analysis
plt.figure(figsize=(8, 5))
sns.histplot(diamonds["price"], bins=40, kde=True)
plt.title("Distribution of Diamond Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()
```

Task 6

Justification:

Histogram is used since price is numeric.
It shows how values are distributed across intervals.
The KDE line also helps visualize the overall shape of the distribution.
(histogram is useful because diamond prices are spread over many values,
and we want to see the overall shape.)

Interpretation:

The distribution of diamond prices is right-skewed.
Most diamonds have lower to medium prices, while less diamonds have very high prices.
The long right tail possible high-price outliers. The center of the data is
closer to the lower price range rather than the maximum.

```python
#Task 6: Categorical variable
plt.figure(figsize=(8, 5))
sns.countplot(data=diamonds, x="cut", order=diamonds["cut"].value_counts().index)
plt.title("Count of Diamonds by Cut")
plt.xlabel("Cut")
plt.ylabel("Count")
plt.show()
```

Justification:

A countplot is suitable because cut is a categorical variable. It shows how many observations belong to each category.

Interpretation:

The most common diamond cut is Ideal, followed by Premium. The least common cut is Fair. The distribution is not uniform because some cut categories appear more often than others.

```python
# Task 7: Bivariate Analysis, Numerical-Numerical
plt.figure(figsize=(8, 5))
sns.scatterplot(data=tips, x="total_bill", y="tip")
plt.title("Relation Between Total Bill and Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

Task 7

Justification:

A scatterplot is suitable because both variables are numerical. It helps show whether there is a relation between two of these variables.

Interpretation:

Total bill and tip are possitively associated. As the total bill increases, the tip also increases. However, the points are not perfectly on a straight line, meaning other factors may also affect the tip amount(outliers).

```python
#Task 7: Numerical-Categorical
plt.figure(figsize=(8, 5))
sns.boxplot(data=tips, x="day", y="total_bill")
plt.title("Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

Task 7: Numerical-Categorical

Justification:

A boxplot is appropriate because day is categorical and total_bill is numerical. It allows us to compare the distribution, median, spread, and outliers of total bills across different days.

(small note: I tried multiple graphs and boxplot was the most suitable)

Interpretation:

Total bills vary by day. Some days have higher median bills and wider spread. Outliers can be seen on some days, meaning there were unusually high bills compared to most bills on those days.

```python
#Task 7: Categorical-Categorical
plt.figure(figsize=(8, 5))
sns.countplot(data=tips, x="time", hue="smoker")
plt.title("Smoker Status by Time")
plt.xlabel("Time")
plt.ylabel("Count")
plt.show()
```

Task 7: Categorical-Categorical

```python
# Task 8: Multivariate Analysis
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=iris,
    x="petal_length",
    y="petal_width",
    hue="species"
)
plt.title("Petal Length vs Petal Width by Species")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.show()
```

Task 8: Iris


Interpretation:


This scatterplot uses three variables:


petal_length

petal_width

species



There are clusters based on species. Setosa is separated from the other species because it has much smaller petal length and petal width. Versicolor and Virginica are closer to each other, but Virginica has larger petal size. Petal length and petal width are positively associated (flowers with longer petals usually also have wider petals).

```python
#Task 8: Tips Dataset — Heatmap
numeric_tips = tips.select_dtypes(include=["int64", "float64"]).corr()
#added the select function by type because the data set has string variable.
plt.figure(figsize=(8, 5))
sns.heatmap(numeric_tips, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix of Numerical Variables in Tips Dataset")
plt.show()
```

Interpretation:

The heatmap shows correlations between all numerical variables in the tips dataset.

The strongest relationship is usually between:

total_bill and tip

This relation is positive, meaning larger bills are generally associated with larger tips.

The variable size also have a positive relation with total_bill, because larger groups usually create larger bills.
